# Wine Quality Analysis – Data Science Tool Mastery
Practice with Pandas, NumPy, Matplotlib and Scikit-learn.

```python
import pandas as pd\nimport numpy as np\nimport matplotlib.pyplot as plt\nfrom sklearn.model_selection import train_test_split\nfrom sklearn.ensemble import RandomForestRegressor\nfrom sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
red = pd.read_csv('data/winequality-red.csv')\nwhite = pd.read_csv('data/winequality-white.csv')\nprint(red.shape, white.shape)\nred.head()
```

## Data quality checks
The original datasets have no missing cells, but contain duplicate records. Raw files are retained; duplicates are removed only in analysis copies.

```python
for name, df in [('Red', red), ('White', white)]:\n    print(name, 'missing:', df.isna().sum().sum(), 'duplicates:', df.duplicated().sum())\nred_clean = red.drop_duplicates().copy()\nwhite_clean = white.drop_duplicates().copy()
```

## Summary statistics

```python
display(red_clean.describe().round(2))\ndisplay(white_clean.describe().round(2))
```

## Visualize quality distribution

```python
plt.hist(red.quality, bins=np.arange(2.5,10.5,1), alpha=.6, label='Red')\nplt.hist(white.quality, bins=np.arange(2.5,10.5,1), alpha=.6, label='White')\nplt.xlabel('Quality'); plt.ylabel('Count'); plt.legend(); plt.show()
```

## Random Forest regression baseline

```python
def evaluate(df, label):\n    X, y = df.drop(columns='quality'), df.quality\n    Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.2, random_state=42)\n    model = RandomForestRegressor(n_estimators=150, random_state=42, n_jobs=-1).fit(Xtr, ytr)\n    pred = model.predict(Xte)\n    print(label, 'MAE', round(mean_absolute_error(yte,pred),3), 'RMSE', round(mean_squared_error(yte,pred)**.5,3), 'R2', round(r2_score(yte,pred),3))\nevaluate(red_clean,'Red')\nevaluate(white_clean,'White')
```
